## Question 1: Isolating One Kind of Symbol

### 📋 Problem Statement
> Count and display the number of capital letters in a given string.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Cypher stares at falling green code and picks out patterns everyone else just sees as noise. This scans a string and counts only the capital letters.
</div>

### 🔄 Workflow
```mermaid
flowchart LR
    A["text"] --> B["for each character"]
    B --> C{"ch.isupper()?"}
    C -->|Yes| D["count += 1"]
    C -->|No| B
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> A generator expression inside `sum()` counts matches without ever building an intermediate list.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
text = "PyThOn LAB 2026"
count_caps = sum(1 for ch in text if ch.isupper())
print("Capital letters:", count_caps)
````

## Question 2: Pulling Out the Vowels

### 📋 Problem Statement
> Count the total number of vowels in a given string.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> The same scan, applied to vowels — isolating one category of character from an otherwise unremarkable sentence.
</div>

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Checking `ch in vowels` against a small string is an easy, readable way to test set membership over a handful of characters.

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
text = "Welcome to Python Programming"
vowels = "aeiouAEIOU"
count_vowels = sum(1 for ch in text if ch in vowels)
print("Total vowels:", count_vowels)
````

## Question 4: Hunting a Recurring Signature

### 📋 Problem Statement
> Enter a string and a substring, then print how many times the substring occurs (left to right).

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Finding every occurrence of a repeated fragment mirrors the crew's hunt for a recurring signature buried inside a longer, noisier signal.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["start = 0"] --> B["idx = str.find(sub, start)"]
    B --> C{"idx == -1?"}
    C -->|Yes| D["stop, return count"]
    C -->|No| E["count += 1"]
    E --> F["start = idx + 1"]
    F --> B
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Advancing `start` by just 1 (not `len(sub)`) after each match correctly catches *overlapping* occurrences.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
main_string = "ABCDCDC"
sub_string = "CDC"

count = 0
start = 0
while True:
    idx = main_string.find(sub_string, start)
    if idx == -1:
        break
    count += 1
    start = idx + 1

print("Occurrences:", count)
````

## Question 5: Tallying Every Letter

### 📋 Problem Statement
> Count occurrences of each alphabet (case-insensitive) in a mixed-case string.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A full frequency count of every letter in a mixed-case string, case-folded so `a` and `A` count as one signal instead of two.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["for ch in text"] --> B{"ch.isalpha()?"}
    B -->|Yes| C["key = ch.upper()"]
    C --> D["frequency[key] += 1"]
    D --> A
    B -->|No| A
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> `dict.get(key, 0)` provides a clean default of zero for a key seen for the first time, avoiding a `KeyError`.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
text = "ABaBCbGc"
frequency = {}

for ch in text:
    if ch.isalpha():
        key = ch.upper()
        frequency[key] = frequency.get(key, 0) + 1

for key in sorted(frequency):
    print(f"{frequency[key]}{key}")
````

## Question 6: Only the Distinct Futures

### 📋 Problem Statement
> Count the number of unique words in a sentence using sets.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Sets strip away duplicates automatically, the way the Oracle only cares about the *distinct* futures that matter, not every repeated variation of the same one.
</div>

### 📐 Formula
$$|\{w_1, w_2, \ldots\}| = \text{count of unique words}$$

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Converting a list of words to a `set` is the idiomatic Python way to deduplicate in a single step.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
sentence = "python is simple and python is powerful"
words = sentence.lower().split()
unique_words = set(words)

print("Unique words:", unique_words)
print("Count of unique words:", len(unique_words))
````

## Question 7: Two Allied Groups

### 📋 Problem Statement
> Create two sets of fruits and find: fruits in both sets, fruits only in `s1`, and the count of all fruits from both sets.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Two separate sets of fruit stand in for two separate groups — intersection finds who's common to both, difference finds who belongs to only one.
</div>

### 📐 Formula
$$S_1 \cap S_2, \qquad S_1 - S_2, \qquad |S_1 \cup S_2|$$

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> `&`, `-`, and `|` on Python sets map directly onto standard mathematical set operations.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
s1 = {"apple", "banana", "mango", "grapes"}
s2 = {"banana", "orange", "kiwi", "mango"}

print("Fruits in both sets:", s1 & s2)
print("Fruits only in s1:", s1 - s2)
print("Count of all unique fruits:", len(s1 | s2))
````

## Question 8: Full Set Algebra

### 📋 Problem Statement
> Take two sets `S1` and `S2` and apply union, intersection, difference, and symmetric difference.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A fuller tour of set algebra — union, intersection, both one-directional differences, and symmetric difference — across two competing factions.
</div>

### 📐 Formula
$$S_1 \cup S_2,\quad S_1 \cap S_2,\quad S_1-S_2,\quad S_2-S_1,\quad S_1 \triangle S_2$$

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> The symmetric difference (`^`) captures everything that's in exactly one set but not both — the 'exclusive or' of set membership.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}

print("Union:", S1 | S2)
print("Intersection:", S1 & S2)
print("Difference (S1 - S2):", S1 - S2)
print("Difference (S2 - S1):", S2 - S1)
print("Symmetric Difference:", S1 ^ S2)
````

## 🌟 Original Extension: Top-3 Most Frequent Words

<div class="alert alert-block alert-info">
<b>💡 Why this extension:</b> Q6 stops at counting *unique* words with a set. This extension goes further and ranks words by how often they actually appear, using a dictionary plus a custom sort key.
</div>

### 🔧 What It Adds

<div class="alert alert-block alert-success">
<b>New capability:</b> Introduces `sorted(..., key=lambda ...)` to rank dictionary items by value — a pattern not used anywhere else in this notebook.
</div>

````python
def most_common_words(text, top_n=3):
    words = text.lower().split()
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1
    ranked = sorted(frequency.items(), key=lambda pair: pair[1], reverse=True)
    return ranked[:top_n]


paragraph = "python is simple python is powerful and python is fun to learn"
print("Top 3 most frequent words:")
for word, count in most_common_words(paragraph):
    print(f"  {word}: {count} time(s)")
````
