## Question 1: The Tally Board

### 📋 Problem Statement
> Scan `n` values in the range 0–3 and print the number of times each value has occurred.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> The Construct keeps count of everything, including how many times a given loadout value has shown up in a scan.
</div>

### 🔄 Workflow
```mermaid
flowchart LR
    A["values list"] --> B["for each value"]
    B --> C{"value in counts?"}
    C -->|Yes| D["counts[value] += 1"]
    D --> B
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> A dictionary pre-seeded with keys `0–3` acts as a running tally board, incremented in a single pass.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
values = [0, 1, 2, 3, 2, 1, 0, 3, 3, 2]
counts = {0: 0, 1: 0, 2: 0, 3: 0}

for value in values:
    if value in counts:
        counts[value] += 1

for key in counts:
    print(f"{key} occurred {counts[key]} time(s)")
````

## Question 2: The Locked-In Record

### 📋 Problem Statement
> Create a tuple of numeric values and find their average.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A tuple — an unchangeable, locked-in sequence — holds a fixed batch of readings, and the program averages them.
</div>

### 📐 Formula
$$\bar{x} = \dfrac{\sum x_i}{n}$$

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Tuples are immutable by design: once a mission log is recorded, it isn't meant to be edited, only analyzed.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
numbers = (12, 25, 30, 18, 15)
average = sum(numbers) / len(numbers)
print("Tuple:", numbers)
print("Average:", average)
````

## Question 3: The Runner-Up

### 📋 Problem Statement
> Given scores of students, print the runner-up score.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Sorting a list of scores and grabbing the second-highest is the Matrix's version of picking the runner-up in a simulated exam.
</div>

### 🔄 Workflow
```mermaid
flowchart LR
    A["scores"] --> B["set(scores)"]
    B --> C["sorted(..., reverse=True)"]
    C --> D["index [1] = runner-up"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Converting to a `set` first removes duplicate scores, so a tie for first place doesn't wrongly become the 'runner-up'.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
scores = [2, 3, 6, 6, 5]
unique_scores = sorted(set(scores), reverse=True)
runner_up = unique_scores[1]
print("Runner-up score:", runner_up)
````

## Question 4: The Crew Registry

### 📋 Problem Statement
> Create a dictionary of persons (name → city) and: display all names, all cities, name-city pairs, and a count of students per city.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A dictionary maps each crew member's name to their location — the kind of registry the Nebuchadnezzar keeps on everyone still connected to the Matrix.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["students dict"] --> B[".keys() → all names"]
    A --> C[".values() → all cities"]
    A --> D[".items() → name, city pairs"]
    D --> E["city_count[city] += 1"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> `.get(city, 0)` lets the city-counter dictionary grow keys on demand without a `KeyError` on the first sighting of a city.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
students = {
    "Aman": "Dehradun",
    "Riya": "Delhi",
    "Vikas": "Dehradun",
    "Neha": "Mumbai",
}

print("All names:", list(students.keys()))
print("All city names:", list(students.values()))

print("\nStudent and city details:")
for name, city in students.items():
    print(f"{name} -> {city}")

city_count = {}
for city in students.values():
    city_count[city] = city_count.get(city, 0) + 1

print("\nCount in each city:")
for city, count in city_count.items():
    print(f"{city}: {count}")
````

## Question 5: The Mission Archive

### 📋 Problem Statement
> Store details of movies (as a list of dictionaries) and: print all details, movies before 2015, profitable movies, and movies by a given director.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A list of dictionaries stands in for the Construct's mission archive — filtered by year, profitability, and director.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["movies list of dicts"] --> B["filter: year < 2015"]
    A --> C["filter: collection > cost"]
    A --> D["filter: director == search_director"]
    B --> E["matching titles"]
    C --> E
    D --> E
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Each filter is an independent pass over the same list — a small preview of the kind of layered querying a real database performs.
</div>

### ⚠️ Watch Out
<div class="alert alert-block alert-warning">
<b>Watch Out:</b> `.lower()` is used for the director-name comparison to make the search case-insensitive — without it, 'raj mehra' would silently fail to match 'Raj Mehra'.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
movies = [
    {"name": "Movie A", "year": 2012, "director": "Raj Mehra", "cost": 50, "collection": 80},
    {"name": "Movie B", "year": 2018, "director": "Anil Verma", "cost": 70, "collection": 65},
    {"name": "Movie C", "year": 2010, "director": "Raj Mehra", "cost": 40, "collection": 55},
]

print("All movie details:")
for movie in movies:
    print(movie)

print("\nMovies released before 2015:")
for movie in movies:
    if movie["year"] < 2015:
        print(movie["name"])

print("\nMovies that made profit:")
for movie in movies:
    if movie["collection"] > movie["cost"]:
        print(movie["name"])

search_director = "Raj Mehra"
print(f"\nMovies directed by {search_director}:")
for movie in movies:
    if movie["director"].lower() == search_director.lower():
        print(movie["name"])
````

## 🌟 Original Extension: Profit Margin Ranking

<div class="alert alert-block alert-info">
<b>💡 Why this extension:</b> Q5 only filters movies into a profitable/not-profitable bucket. This extension computes an actual profit-margin percentage per movie and ranks the whole list by it.
</div>

### 🔧 What It Adds

<div class="alert alert-block alert-success">
<b>New capability:</b> Adds a derived metric function and sorts a list of dictionaries by a computed (not stored) value — a step beyond simple filtering.
</div>

````python
movies = [
    {"name": "Movie A", "year": 2012, "director": "Raj Mehra", "cost": 50, "collection": 80},
    {"name": "Movie B", "year": 2018, "director": "Anil Verma", "cost": 70, "collection": 65},
    {"name": "Movie C", "year": 2010, "director": "Raj Mehra", "cost": 40, "collection": 55},
]


def profit_margin(movie):
    return round(((movie["collection"] - movie["cost"]) / movie["cost"]) * 100, 1)


ranked_movies = sorted(movies, key=profit_margin, reverse=True)

print("Movies ranked by profit margin (%):")
for movie in ranked_movies:
    print(f"  {movie['name']}: {profit_margin(movie)}%")
````
