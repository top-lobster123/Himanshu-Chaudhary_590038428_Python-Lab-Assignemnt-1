## Question 1: The Repeating Motion

### 📋 Problem Statement
> Find the factorial of a given number using a loop.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Neo's mastery doesn't come from one download — it comes from the same operation looping over and over until it becomes instinct. Factorial multiplies, repeats, multiplies again.
</div>

### 📐 Formula
$$n! = \prod_{i=1}^{n} i$$

### 🔄 Workflow
```mermaid
flowchart TD
    A["factorial = 1"] --> B["for i in 1..num"]
    B --> C["factorial *= i"]
    C --> B
    B -->|loop done| D["return factorial"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> A `for` loop turns a mathematical product into a running accumulator — five lines instead of a wall of repeated multiplication.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 5
factorial = 1

for i in range(1, num + 1):
    factorial *= i

print(f"Factorial of {num} is {factorial}")
````

## Question 2: The Hidden Signature

### 📋 Problem Statement
> Check whether the given number is an Armstrong number.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Armstrong numbers are the Matrix's version of a hidden signature — a number that secretly reconstructs itself when each digit is raised to the right power.
</div>

### 📐 Formula
$$\sum_{i} d_i^{\,k} = n, \quad k = \text{number of digits in } n$$

### 🔄 Workflow
```mermaid
flowchart TD
    A["num as string"] --> B["for each digit"]
    B --> C["sum += digit ** power"]
    C --> B
    B -->|done| D{"sum == num?"}
    D -->|Yes| E["Armstrong number"]
    D -->|No| F["Not Armstrong"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> The digit count (`power`) must be computed once before the loop starts — it's the exponent every digit gets raised to.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 153
num_str = str(num)
power = len(num_str)
sum_of_powers = 0

for digit in num_str:
    sum_of_powers += int(digit) ** power

if sum_of_powers == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
````

## Question 3: Nature's Favorite Spiral

### 📋 Problem Statement
> Print the Fibonacci series up to a given number of terms.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> The Fibonacci sequence is exactly the kind of self-referencing pattern that feels like it shouldn't be simulate-able — yet a loop generates it term by term.
</div>

### 📐 Formula
$$F_n = F_{n-1} + F_{n-2}, \qquad F_0=0,\ F_1=1$$

### 🔄 Workflow
```mermaid
flowchart LR
    A["a=0, b=1"] --> B["print a"]
    B --> C["a, b = b, a+b"]
    C --> B
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Simultaneous tuple assignment (`a, b = b, a+b`) advances both variables in one atomic step — no temp variable needed.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
terms = 10
a, b = 0, 1

print("Fibonacci series:")
for _ in range(terms):
    print(a, end=" ")
    a, b = b, a + b
print()
````

## Question 4: Hunting for Hidden Factors

### 📋 Problem Statement
> Check whether a given number is prime or not.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Primality testing is the machine checking a number for hidden factors, dividing by everything up to its square root before declaring it clean.
</div>

### 📐 Formula
$$\text{prime}(n) \iff \nexists\, i \in [2,\sqrt{n}]\ \text{s.t. } n \bmod i = 0$$

### 🔄 Workflow
```mermaid
flowchart TD
    A["num"] --> B{"num < 2?"}
    B -->|Yes| F["Not prime"]
    B -->|No| C["for i in 2..sqrt(num)"]
    C --> D{"num % i == 0?"}
    D -->|Yes| E["Not prime, break"]
    D -->|No| C
    C -->|loop done| G["Prime"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Only divisors up to `√num` need checking — any factor larger than that would have a matching factor smaller than it, already tested.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 29

if num < 2:
    print(num, "is not prime")
else:
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    print(num, "is prime" if is_prime else "is not prime")
````

## Question 5: Deja Vu Made Literal

### 📋 Problem Statement
> Check whether a given number is a palindrome or not.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A palindrome reads the same forwards and backwards — deja vu made literal, caught by reversing the digits and comparing.
</div>

### 🔄 Workflow
```mermaid
flowchart LR
    A["num"] --> B["str(num)"]
    B --> C["reversed = str(num)[::-1]"]
    C --> D{"str(num) == reversed?"}
    D -->|Yes| E["Palindrome"]
    D -->|No| F["Not a palindrome"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Python's slice trick `[::-1]` reverses a string in a single expression — no explicit loop required for the reversal itself.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 12321

if str(num) == str(num)[::-1]:
    print(num, "is a palindrome")
else:
    print(num, "is not a palindrome")
````

## Question 6: Breaking a Number Into Pieces

### 📋 Problem Statement
> Print the sum of digits of a number.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Digit by digit, this loop breaks a number down into its individual pieces and adds them back up.
</div>

### 🔄 Workflow
```mermaid
flowchart LR
    A["num as string"] --> B["for each digit"]
    B --> C["sum_digits += int(digit)"]
    C --> B
    B -->|done| D["print sum_digits"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Converting the number to a string first makes each digit trivially iterable — a common Python pattern for digit-level work.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 9876
sum_digits = 0

for digit in str(num):
    sum_digits += int(digit)

print("Sum of digits:", sum_digits)
````

## Question 7: Filtering the Range

### 📋 Problem Statement
> Count and print all numbers divisible by 5 or 7 between 1 and 100.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A sweep from 1 to 100, flagging anything divisible by 5 or 7 — the same kind of pattern-matching sweep Tank runs across sonar data.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["n in 1..100"] --> B{"n % 5 == 0 OR n % 7 == 0?"}
    B -->|Yes| C["keep n"]
    B -->|No| D["discard n"]
    C --> E["numbers list"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> A list comprehension expresses the entire filter-loop in a single readable line.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
numbers = [n for n in range(1, 101) if n % 5 == 0 or n % 7 == 0]
print("Numbers:", numbers)
print("Count:", len(numbers))
````

## Question 9: Cataloguing Every Clean Signal

### 📋 Problem Statement
> Print all prime numbers between 1 and 100.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> The same primality check from before, run across an entire range — a nested loop sweeping 1 to 100 and collecting every prime it finds.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["num in 2..100"] --> B["inner: test divisors up to sqrt(num)"]
    B --> C{"prime?"}
    C -->|Yes| D["append to primes list"]
    C -->|No| A
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> This is O(n√n) overall — fine for a small range like 1–100, but a Sieve of Eratosthenes would scale far better for larger ranges.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
primes = []

for num in range(2, 101):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(num)

print("Prime numbers from 1 to 100:")
print(primes)
````

## Question 10: The Drilling Table

### 📋 Problem Statement
> Print the multiplication table for a given number.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A multiplication table is the simplest possible training loop: run the same operation ten times with an incrementing counter.
</div>

### 📐 Formula
$$num \times i,\quad i = 1 \ldots 10$$

### 🔄 Workflow
```mermaid
flowchart LR
    A["i = 1..10"] --> B["print(num * i)"]
    B --> A
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> A single `for i in range(1, 11)` replaces ten nearly-identical print statements.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 5
for i in range(1, 11):
    print(f"{num} * {i} = {num * i}")
````

## 🌟 Original Extension: Twin Prime Finder

<div class="alert alert-block alert-info">
<b>💡 Why this extension:</b> Q4 and Q9 test primality one number at a time. This extension refactors that logic into a reusable `is_prime()` function and uses it to solve a genuinely new problem: finding twin primes (pairs that differ by exactly 2).
</div>

### 🔧 What It Adds

<div class="alert alert-block alert-success">
<b>New capability:</b> Refactors inline primality-check code into a reusable function, then composes it with a list comprehension to solve a problem not in the original ten.
</div>

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


twin_primes = [(n, n + 2) for n in range(2, 100) if is_prime(n) and is_prime(n + 2)]
print("Twin prime pairs between 1 and 100:")
print(twin_primes)
````
