## Question 8: Area Without an Angle

### 📋 Problem Statement
> Find the area of a triangle when the lengths of all three sides are given (Heron's formula).

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Given only three side-lengths, Heron's formula lets the machine compute area without ever needing an angle.
</div>

### 📐 Formula
$$s = \dfrac{a+b+c}{2}, \qquad Area = \sqrt{s(s-a)(s-b)(s-c)}$$

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> The semi-perimeter `s` is computed first and reused three times inside the square root.
</div>

### ⚠️ Watch Out
<div class="alert alert-block alert-warning">
<b>Watch Out:</b> If `a`, `b`, `c` can't actually form a triangle, the expression under the square root goes negative — always validate the triangle inequality first in production code.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
import math

a, b, c = 3, 4, 5
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print("Area of triangle:", area)
````

## Question 9: Reversing a Countdown

### 📋 Problem Statement
> Convert a given number of seconds into hours, minutes, and remaining seconds.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Tank counts down mission time before Neo jacks out. This program reverses that from raw seconds into hours, minutes, and seconds.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["total_seconds"] --> B["hours = total_seconds // 3600"]
    B --> C["remainder = total_seconds % 3600"]
    C --> D["minutes = remainder // 60"]
    D --> E["seconds = remainder % 60"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Integer division (`//`) peels off whole units, and modulus (`%`) hands back exactly what's left for the next step.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
total_seconds = 7384

hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60

print(f"{hours} hour(s), {minutes} minute(s), {seconds} second(s)")
````

## Question 10: The Temp-Free Swap

### 📋 Problem Statement
> Swap two numbers without using an additional variable.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Trinity swaps two things without a spare hand to hold one — same principle here: swapping `a` and `b` using pure arithmetic, no extra variable.
</div>

### 🔄 Workflow
```mermaid
flowchart LR
    S1["a=25, b=40"] --> S2["a = a + b"]
    S2 --> S3["b = a - b → original a"]
    S3 --> S4["a = a - b → original b"]
    S4 --> S5["a=40, b=25"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Three arithmetic reassignments fully swap two values without ever introducing a third variable.
</div>

### ⚠️ Watch Out
<div class="alert alert-block alert-warning">
<b>Watch Out:</b> This trick can silently overflow in fixed-width languages; Python's arbitrary-precision integers make it safe here, but `a, b = b, a` is still the more idiomatic Python swap.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
a = 25
b = 40

print("Before swap:", a, b)
a = a + b
b = a - b
a = a - b
print("After swap:", a, b)
````

## Question 11: The Gauss Shortcut

### 📋 Problem Statement
> Find the sum of the first `n` natural numbers.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> A classic shortcut: summing every natural number from 1 to `n` without looping through them one by one.
</div>

### 📐 Formula
$$\sum_{i=1}^{n} i = \dfrac{n(n+1)}{2}$$

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> A closed-form formula replaces an entire loop — O(1) instead of O(n).
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
n = 10
sum_n = n * (n + 1) // 2
print("Sum of first", n, "natural numbers is", sum_n)
````

## Question 13: Sliding the Code Sideways

### 📋 Problem Statement
> Find the left-shift and right-shift values of a given number.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Bit-shifting a number is like sliding its binary representation sideways — each shift left doubles the value, each shift right halves it.
</div>

### 📐 Formula
$$n \ll k = n \times 2^{k}, \qquad n \gg k = \left\lfloor \dfrac{n}{2^{k}} \right\rfloor$$

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Shifting is a fast, hardware-level way to multiply or divide by powers of two.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 10
shift = 2

print("Number:", num)
print("Left shift by 2:", num << shift)
print("Right shift by 2:", num >> shift)
````

## 🌟 Original Extension: Simple vs. Compound Interest Comparator

<div class="alert alert-block alert-info">
<b>💡 Why this extension:</b> The base experiment only computes simple interest. Real financial decisions inside (and outside) the simulation usually need to compare it against compound interest — so this extension adds a second function and a direct comparison.
</div>

### 🔧 What It Adds

<div class="alert alert-block alert-success">
<b>New capability:</b> Introduces `**` for exponentiation, a second reusable function, and a derived comparison metric not requested anywhere in the original 15 questions.
</div>

````python
def simple_interest(principal, rate, time):
    return (principal * rate * time) / 100


def compound_interest(principal, rate, time):
    return principal * ((1 + rate / 100) ** time) - principal


principal, rate, time = 10000, 8, 2
si = simple_interest(principal, rate, time)
ci = compound_interest(principal, rate, time)

print(f"Simple Interest  : {si:.2f}")
print(f"Compound Interest: {ci:.2f}")
print(f"Compounding earns {ci - si:.2f} more over {time} year(s).")
````
