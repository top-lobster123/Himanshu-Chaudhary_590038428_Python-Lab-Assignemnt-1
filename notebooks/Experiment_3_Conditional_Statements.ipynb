## Question 1: Passing a Double Gate

### 📋 Problem Statement
> Check whether a given number is divisible by both 3 and 5.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Every meaningful choice in the Matrix is really several conditions layered together — divisible by 3 **and** by 5, both gates having to pass at once.
</div>

### 📐 Formula
$$n \bmod 3 = 0 \ \land\ n \bmod 5 = 0$$

### 🔄 Workflow
```mermaid
flowchart TD
    A["num"] --> B{"num % 3 == 0 AND num % 5 == 0?"}
    B -->|True| C["divisible by both"]
    B -->|False| D["not divisible by both"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> `and` requires both sub-conditions to be `True` before the branch is taken — a compound gate, not two separate checks.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
num = 45

if num % 3 == 0 and num % 5 == 0:
    print(num, "is divisible by both 3 and 5")
else:
    print(num, "is not divisible by both 3 and 5")
````

## Question 3: Which Is Greater?

### 📋 Problem Statement
> Find the greatest among two numbers. If equal, print 'numbers are equal'.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> "Which is greater?" is the oldest branching question there is — and the Architect refuses to force a false winner when the values are actually equal.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["a, b"] --> B{"a > b?"}
    B -->|Yes| C["Greatest = a"]
    B -->|No| D{"b > a?"}
    D -->|Yes| E["Greatest = b"]
    D -->|No| F["numbers are equal"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> `elif` chains let the code check a second possibility only after the first one fails, ending in a fallback `else`.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
a = 15
b = 15

if a > b:
    print("Greatest number is", a)
elif b > a:
    print("Greatest number is", b)
else:
    print("numbers are equal")
````

## Question 4: Three-Way Comparison

### 📋 Problem Statement
> Find the greatest among three numbers (assuming all values are different).

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> The same comparison, scaled to three values at once — a taste of the layered `elif` chains Neo will need once choices stop being simple pairs.
</div>

### 🔄 Workflow
```mermaid
flowchart TD
    A["a, b, c"] --> B{"a>b AND a>c?"}
    B -->|Yes| G["Greatest = a"]
    B -->|No| C{"b>a AND b>c?"}
    C -->|Yes| H["Greatest = b"]
    C -->|No| I["Greatest = c"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Because the values are guaranteed distinct, once the first two comparisons fail the third value must be the largest by elimination.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
a, b, c = 12, 45, 32

if a > b and a > c:
    print("Greatest number is", a)
elif b > a and b > c:
    print("Greatest number is", b)
else:
    print("Greatest number is", c)
````

## Question 5: Real or Imaginary?

### 📋 Problem Statement
> Check whether a quadratic equation `ax^2 + bx + c = 0` has real or imaginary roots and display the roots.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Not every equation resolves cleanly — some roots exist only in the space between what's possible and what isn't. The discriminant decides which kind of root the equation produces.
</div>

### 📐 Formula
$$D = b^{2}-4ac, \qquad x = \dfrac{-b \pm \sqrt{D}}{2a}$$

### 🔄 Workflow
```mermaid
flowchart TD
    A["a, b, c"] --> B["D = b² − 4ac"]
    B --> C{"D >= 0?"}
    C -->|Yes| D1["Roots are real"]
    C -->|No| D2["Roots are imaginary"]
    D1 --> E["compute via cmath.sqrt"]
    D2 --> E
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> `cmath.sqrt` (instead of `math.sqrt`) safely handles the negative-discriminant case and returns a complex number instead of crashing.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
import cmath

a, b, c = 1, 2, 5

discriminant = b ** 2 - 4 * a * c
root1 = (-b + cmath.sqrt(discriminant)) / (2 * a)
root2 = (-b - cmath.sqrt(discriminant)) / (2 * a)

if discriminant >= 0:
    print("Roots are real")
else:
    print("Roots are imaginary")

print("Root 1:", root1)
print("Root 2:", root2)
````

## Question 6: Faking a Believable Calendar

### 📋 Problem Statement
> Find whether a given year is a leap year or not.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> Even the Matrix needs a convincing calendar. Leap years follow their own nested rule so 1999 never feels fake.
</div>

### 📐 Formula
$$(year \bmod 400 = 0) \ \lor\ (year \bmod 4 = 0 \ \land\ year \bmod 100 \neq 0)$$

### 🔄 Workflow
```mermaid
flowchart TD
    A["year"] --> B{"year % 400 == 0?"}
    B -->|Yes| L["Leap year"]
    B -->|No| C{"year % 4 == 0 AND year % 100 != 0?"}
    C -->|Yes| L
    C -->|No| N["Not a leap year"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Century years (like 1900) are the tricky edge case: divisible by 4 but *not* a leap year unless also divisible by 400.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
year = 2024

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
````

> Take any date as input and display the next date in the calendar.

<div class="alert alert-block alert-info">

## Question 8: The Full Gradesheet

### 📋 Problem Statement
> Print the grade sheet of a student from marks of five subjects, computing percentage, CGPA, and grade.

<div class="alert alert-block alert-info">
<b>🧠 The Matrix Angle:</b> The Architect judges Neo not by feelings but by hard thresholds — and so does this grade calculator, routing a CGPA through a ladder of conditions.
</div>

### 📐 Formula
$$Percentage = \dfrac{\sum marks}{500}\times100, \qquad CGPA = \dfrac{Percentage}{10}$$

### 🔄 Workflow
```mermaid
flowchart TD
    A["CGPA"] --> B{"0.0–3.4?"}
    B -->|Yes| G1["F"]
    B -->|No| C{"3.5–5.0?"}
    C -->|Yes| G2["C+"]
    C -->|No| D{"5.1–6.0?"}
    D -->|Yes| G3["B"]
    D -->|No| E{"6.1–7.0?"}
    E -->|Yes| G4["B+"]
    E -->|No| F{"7.1–8.0?"}
    F -->|Yes| G5["A"]
    F -->|No| H{"8.1–9.0?"}
    H -->|Yes| G6["A+"]
    H -->|No| I{"9.1–10.0?"}
    I -->|Yes| G7["O (Outstanding)"]
    I -->|No| G8["Invalid CGPA"]
```

### ✅ Key Takeaway
<div class="alert alert-block alert-success">
<b>Key Takeaway:</b> Each CGPA range is checked top-to-bottom until exactly one matches — the same kind of strict, ordered rule-bound evaluation every choice inside the Matrix reduces to.
</div>

### ⚠️ Watch Out
<div class="alert alert-block alert-warning">
<b>Watch Out:</b> The boundaries are defined with fixed decimal cutoffs (e.g. 3.4 vs 3.5) — a CGPA that rounds unexpectedly (like 3.45 becoming 3.5) can quietly land in a different bracket than intended.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
def get_grade(cgpa):
    if 0.0 <= cgpa <= 3.4:
        return "F"
    if 3.5 <= cgpa <= 5.0:
        return "C+"
    if 5.1 <= cgpa <= 6.0:
        return "B"
    if 6.1 <= cgpa <= 7.0:
        return "B+"
    if 7.1 <= cgpa <= 8.0:
        return "A"
    if 8.1 <= cgpa <= 9.0:
        return "A+"
    if 9.1 <= cgpa <= 10.0:
        return "O (Outstanding)"
    return "Invalid CGPA"

name = "Rohit Sharma"
roll_no = "R17234512"
sap_id = "50005673"
semester = 1
course = "B.Tech. CSE AI & ML"

marks = {
    "PDS": 70,
    "Python": 80,
    "Chemistry": 90,
    "English": 60,
    "Physics": 65,
}

total = sum(marks.values())
percentage = (total / (len(marks) * 100)) * 100
cgpa = round(percentage / 10, 1)
grade = get_grade(cgpa)

print("Sample Gradesheet")
print("Name:", name)
print("Roll Number:", roll_no, "SAPID:", sap_id)
print("Sem:", semester, "Course:", course)
print("Subject name: Marks")
for subject, score in marks.items():
    print(f"{subject}: {score}")
print(f"Percentage: {percentage:.2f}%")
print("CGPA:", cgpa)
print("Grade:", grade)
````

## 🌟 Original Extension: Days-in-Month Calculator

<div class="alert alert-block alert-info">
<b>💡 Why this extension:</b> Q6 only checks whether a year is a leap year. This extension reuses that exact rule inside a more general function that reports how many days are in *any* month of *any* year.
</div>

### 🔧 What It Adds

<div class="alert alert-block alert-success">
<b>New capability:</b> Generalizes a single-purpose leap-year check into a reusable `days_in_month()` function, tested across several month/year combinations at once.
</div>

````python
def days_in_month(month, year):
    thirty_day_months = (4, 6, 9, 11)
    if month == 2:
        is_leap = (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)
        return 29 if is_leap else 28
    elif month in thirty_day_months:
        return 30
    else:
        return 31


test_cases = [(2, 2024), (2, 1900), (4, 2025), (12, 2025)]
for month, year in test_cases:
    print(f"Month {month}, Year {year} -> {days_in_month(month, year)} days")
````
