### ⚠️ Watch Out
<div class="alert alert-block alert-warning">
<b>Watch Out:</b> `"12\05\1999"` looks like a date, but `\0` is quietly consumed as an escape attempt. Prefer raw strings (`r"..."`) whenever backslashes aren't meant to be escape codes.
</div>

*Sample input/output for this problem is shown directly below, in the executed code cell's output — this notebook has already been run once end-to-end.*

````python
# 2(a)
print("Hello Everyone !!!")

# 2(b)
print("Hello")
print("World")

# 2(c)
print("Hello\nWorld")

# 2(d)
print("Rohit's date of birth is 12\05\1999")
````

## 🌟 Original Extension: The Crew ID Card Generator

<div class="alert alert-block alert-info">
<b>💡 Why this extension:</b> The original Q7 just prints personal fields one by one. Real crew dossiers on the Nebuchadnezzar are formatted, bordered records — so this extension wraps the same fields in a reusable function that builds a clean, aligned ID card.
</div>

### 🔧 What It Adds

<div class="alert alert-block alert-success">
<b>New capability:</b> Turns repeated `print()` calls into one reusable function, and introduces f-string alignment specifiers (`:^40`) that weren't used anywhere in the base experiment.
</div>

````python
def generate_id_card(name, sap_id, dob, programme, semester):
    """Builds a formatted, presentation-ready ID card from raw personal fields."""
    border = "=" * 40
    card = (
        f"{border}\n"
        f"{'NEBUCHADNEZZAR CREW ID CARD':^40}\n"
        f"{border}\n"
        f"Name       : {name}\n"
        f"SAP ID     : {sap_id}\n"
        f"DOB        : {dob}\n"
        f"Programme  : {programme}\n"
        f"Semester   : {semester}\n"
        f"{border}"
    )
    return card


print(generate_id_card("Nikunj Bansal", "500069944", "13 Oct 1999", "AI & ML", 2))
````
